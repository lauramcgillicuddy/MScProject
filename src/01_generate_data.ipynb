# 01 — Generate Grammar Data

Generates training strings from a grammar definition file and saves them to disk.

Run this notebook for each grammar before training. Available grammars are in `grammars/`:
- `anbn.txt` — aⁿbⁿ (context-free)
- `cndn.txt` — cⁿdⁿ (context-free)
- `anbncn.txt` — aⁿbⁿcⁿ (context-sensitive, direct pattern)
- `hierarchical.txt` — nested bracketing structure

```python
from pathlib import Path
from grammar_loader import load_grammar, generate_strings
```

## Configuration

Edit the cells below to choose which grammar to use and the range of `n`.

```python
GRAMMAR_FILE = 'grammars/anbn.txt'   # path to grammar definition
MIN_N        = 1                      # smallest n to generate
MAX_N        = 100                    # largest n to generate (Barry: use up to 100)
OUTPUT_FILE  = None                   # None => auto: data/<name>_n<min>-<max>.txt
```

## Load grammar and generate strings

```python
grammar = load_grammar(GRAMMAR_FILE)
strings = generate_strings(grammar, min_n=MIN_N, max_n=MAX_N)

print(f"Grammar  : {grammar.name}  ({grammar.gtype})")
print(f"Terminals: {grammar.terminals}")
print(f"Strings  : {len(strings)}  (n={MIN_N}..{MAX_N})")
print(f"Lengths  : {min(len(s) for s in strings)} .. {max(len(s) for s in strings)} chars")
print()
print("First 5 strings:")
for s in strings[:5]:
    print(f"  {s!r}")
```

## Save to disk

```python
output_path = OUTPUT_FILE or f'data/{grammar.name}_n{MIN_N}-{MAX_N}.txt'
Path(output_path).parent.mkdir(parents=True, exist_ok=True)

with open(output_path, 'w') as f:
    for s in strings:
        f.write(s + '\n')

print(f"Saved {len(strings)} strings to: {output_path}")
```
